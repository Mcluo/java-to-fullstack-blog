## 1. 变量和类型

Python 的变量不需要声明类型，类似 Java 的 `var` 关键字。

```python
# Python 写法
name = "Java工程师"
age = 30
salary = 25000.50
is_employed = True

print(f"姓名: {name}")
print(f"年龄: {age}")
print(f"薪资: {salary}")
print(f"在职: {is_employed}")
```

```python
# 类型检查
print(f"name 的类型: {type(name)}")
print(f"age 的类型: {type(age)}")
print(f"salary 的类型: {type(salary)}")
```

## 2. 列表（类似 Java 的 ArrayList）

```python
# 创建列表
skills = ["Java", "Spring", "MySQL", "Redis"]

# 添加元素
skills.append("Python")
skills.append("Django")

print("技能列表:", skills)
print("第一个技能:", skills[0])
print("最后一个技能:", skills[-1])
```

```python
# 列表切片
print("前3个技能:", skills[:3])
print("后3个技能:", skills[-3:])
print("中间技能:", skills[2:4])
```

## 3. 字典（类似 Java 的 HashMap）

```python
# 创建字典
person = {
    "name": "张三",
    "age": 30,
    "skills": ["Java", "Python"],
    "experience": 5
}

print("姓名:", person["name"])
print("年龄:", person["age"])
print("技能:", person["skills"])
```

```python
# 遍历字典
for key, value in person.items():
    print(f"{key}: {value}")
```

## 4. 函数定义

```python
# 定义函数
def calculate_salary(base_salary, bonus_rate=0.2):
    """
    计算总薪资
    
    参数:
        base_salary: 基本工资
        bonus_rate: 奖金比例（默认20%）
    
    返回:
        总薪资
    """
    bonus = base_salary * bonus_rate
    total = base_salary + bonus
    return total

# 调用函数
print(f"总薪资: {calculate_salary(20000)}")
print(f"总薪资（30%奖金）: {calculate_salary(20000, 0.3)}")
```

## 5. 类和对象

```python
# 定义类
class Employee:
    def __init__(self, name, age, position):
        self.name = name
        self.age = age
        self.position = position
    
    def introduce(self):
        return f"我是{self.name}，{self.age}岁，职位是{self.position}"
    
    def birthday(self):
        self.age += 1
        return f"{self.name} 过生日了，现在{self.age}岁"

# 创建对象
emp = Employee("李四", 28, "Python工程师")
print(emp.introduce())
print(emp.birthday())
```
